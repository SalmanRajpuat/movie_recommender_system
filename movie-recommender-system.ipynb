```python
#as it is string of lists so we will convert it into list so we can apply the convert function that we made
import ast 
```

```python
#here literal eval from ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
features['genres'] = features['genres'].apply(convert) 
```

```python
features['keywords'] = features['keywords'].apply(convert)
```

```python
features.head()
```

```python
features['cast'][0] 
```

```python
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter!=3:
            L.append(i['name'])
            counter+=1
        else:
            break;
    return L
```

```python
features['cast'] = features['cast'].apply(convert3)
```

```python
features.head()
```

```python
features['crew'][0]
```

```python
def fetchDirector(obj):
    L = []
    
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
features['crew']= features['crew'].apply(fetchDirector)
```

```python
features.head()
```

```python
features['overview'] = features['overview'].apply(lambda x:x.split())
```

```python
features.head()
```

```python
features['genres'] = features['genres'].apply(lambda x:[i.replace(" ", "") for i in x])
features['keywords'] = features['keywords'].apply(lambda x:[i.replace(" ", "") for i in x])
features['cast'] = features['cast'].apply(lambda x:[i.replace(" ", "") for i in x])
features['crew'] = features['crew'].apply(lambda x:[i.replace(" ", "") for i in x])
```

```python
features.head()
```

```python
features['tags'] = features['overview'] + features['genres'] + features['keywords'] + features['cast'] + features['crew']
```

```python
new_df = features[['movie_id', 'title', 'tags']]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=10000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
import nltk 
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]
    
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
 cosine_similarity(vectors).shape
```

```python
import difflib
```

```python
def recommend(movie):
    
    # Get closest match from title list
    close_matches = difflib.get_close_matches(movie, new_df['title'].values)
    
    if not close_matches:
        print("Movie not found.")
        return
    
    movie = close_matches[0]  # Best match
    
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    
    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:9]
    
    print("Showing results for:", movie)
    print("------------------------")
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Batman')
```
